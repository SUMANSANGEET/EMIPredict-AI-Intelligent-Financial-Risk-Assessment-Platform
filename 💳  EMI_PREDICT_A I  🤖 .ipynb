#  🧹 Data Cleaning pipeline — the raw file has realistic data-quality issues ⚙️ : 

- **🔢 MIXED TYPES - age / monthly_salary / bank_balance stored as mixed str/float (e.g. '38' and 38.0)**

- **👤 INCONSISTENCY  - gender has inconsistent casing/abbreviations (Male/MALE/male/M/F/Female/female/FEMALE)**

- **🕳️ MISSING DATA - education has missing values**


````python
before = len(df)
df = df.drop_duplicates()
print(f"Removed {before - len(df):,} duplicate rows -> {len(df):,} rows remain")

# 1) Coerce numeric-as-string columns to true numeric dtype
for c in ['age', 'monthly_salary', 'bank_balance']:
    df[c] = pd.to_numeric(df[c], errors='coerce')

# 2) Standardize inconsistent categorical labels
gender_map = {'male':'Male','MALE':'Male','M':'Male','Male':'Male',
              'female':'Female','FEMALE':'Female','F':'Female','Female':'Female'}
df['gender'] = df['gender'].map(gender_map)

num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = df.select_dtypes(include=['object']).columns.tolist()

print("\nMissing values introduced/found before imputation:")
print(df[num_cols + cat_cols].isnull().sum().pipe(lambda s: s[s>0]))

for c in num_cols:
    if df[c].isnull().any():
        df[c] = df[c].fillna(df[c].median())
for c in cat_cols:
    if df[c].isnull().any():
        df[c] = df[c].fillna(df[c].mode()[0])

# 3) Sanity clip on domain bounds
df['credit_score'] = df['credit_score'].clip(300, 850)
df['age'] = df['age'].clip(18, 75)

print("\nPost-cleaning missing values:", df.isnull().sum().sum())
print("Post-cleaning gender categories:", df['gender'].unique().tolist())
print("Final shape:", df.shape)
````

## ⚙️ Step 3 — Feature Engineering

````python

def engineer_features(data):
    d = data.copy()
    total_expenses = (d['monthly_rent'] + d['school_fees'] + d['college_fees'] + d['travel_expenses'] +
                       d['groceries_utilities'] + d['other_monthly_expenses'] + d['current_emi_amount'])
    d['total_monthly_expenses'] = total_expenses
    d['debt_to_income_ratio'] = d['current_emi_amount'] / d['monthly_salary'].replace(0, np.nan)
    d['expense_to_income_ratio'] = total_expenses / d['monthly_salary'].replace(0, np.nan)
    d['disposable_income'] = d['monthly_salary'] - total_expenses
    d['affordability_ratio'] = d['disposable_income'] / d['monthly_salary'].replace(0, np.nan)
    d['savings_rate'] = d['bank_balance'] / d['monthly_salary'].replace(0, np.nan)
    d['emergency_coverage_months'] = d['emergency_fund'] / total_expenses.replace(0, np.nan)
    d['requested_to_income_ratio'] = d['requested_amount'] / (d['monthly_salary'] * d['requested_tenure']).replace(0, np.nan)
    d['dependents_per_earner'] = d['dependents'] / 1.0
    d['employment_stability_score'] = d['years_of_employment'] * (d['employment_type']=='Government').map({True:1.3, False:1.0})
    d['credit_risk_score'] = (d['credit_score'] - 300) / (850 - 300)  # normalized 0-1
    d = d.replace([np.inf, -np.inf], np.nan)
    fe_cols = ['total_monthly_expenses','debt_to_income_ratio','expense_to_income_ratio','disposable_income',
               'affordability_ratio','savings_rate','emergency_coverage_months','requested_to_income_ratio',
               'dependents_per_earner','employment_stability_score','credit_risk_score']
    for c in fe_cols:
        d[c] = d[c].fillna(d[c].median())
    return d

train_fe = engineer_features(train_df)
val_fe = engineer_features(val_df)
test_fe = engineer_features(test_df)
print("Engineered feature columns added:")
print(['total_monthly_expenses','debt_to_income_ratio','expense_to_income_ratio','disposable_income',
       'affordability_ratio','savings_rate','emergency_coverage_months','requested_to_income_ratio',
       'dependents_per_earner','employment_stability_score','credit_risk_score'])
train_fe.filter(like='ratio').describe().T
````

# ⚙️ Categorical encoding (LabelEncoder fit on train, applied everywhere) + scaling 📊

````python
from sklearn.preprocessing import LabelEncoder, StandardScaler
import joblib

# Categorical and numerical features
cat_features = [
    'gender', 'marital_status', 'education', 'employment_type',
    'company_type', 'house_type', 'existing_loans', 'emi_scenario'
]

num_features = [
    'age', 'monthly_salary', 'years_of_employment', 'monthly_rent',
    'family_size', 'dependents', 'school_fees', 'college_fees',
    'travel_expenses', 'groceries_utilities', 'other_monthly_expenses',
    'current_emi_amount', 'credit_score', 'bank_balance',
    'emergency_fund', 'requested_amount', 'requested_tenure',
    'total_monthly_expenses', 'debt_to_income_ratio',
    'expense_to_income_ratio', 'disposable_income',
    'affordability_ratio', 'savings_rate',
    'emergency_coverage_months', 'requested_to_income_ratio',
    'dependents_per_earner', 'employment_stability_score',
    'credit_risk_score'
]

# ==========================================
# FAST CATEGORICAL ENCODING
# ==========================================

encoders = {}

for c in cat_features:
    
    le = LabelEncoder()
    
    # Fit only on training data
    train_fe[c] = train_fe[c].astype(str)
    le.fit(train_fe[c])
    
    encoders[c] = le
    
    # Encode training data
    train_fe[c + '_enc'] = le.transform(train_fe[c])
    
    # Encode validation and test data safely
    mapping = {label: idx for idx, label in enumerate(le.classes_)}
    
    val_fe[c + '_enc'] = (
        val_fe[c].astype(str).map(mapping).fillna(-1).astype(int)
    )
    
    test_fe[c + '_enc'] = (
        test_fe[c].astype(str).map(mapping).fillna(-1).astype(int)
    )


# ==========================================
# FEATURE LIST
# ==========================================

enc_features = [c + '_enc' for c in cat_features]

feature_cols = num_features + enc_features


# ==========================================
# FAST SCALING
# ==========================================

scaler = StandardScaler()

# Fit scaler on training data only
scaler.fit(train_fe[feature_cols])

scaled_cols = [c + '_scaled' for c in feature_cols]

# Transform all datasets
train_fe[scaled_cols] = scaler.transform(train_fe[feature_cols])
val_fe[scaled_cols] = scaler.transform(val_fe[feature_cols])
test_fe[scaled_cols] = scaler.transform(test_fe[feature_cols])


# ==========================================
# SAVE MODELS
# ==========================================

joblib.dump(encoders, 'models/label_encoders.pkl')
joblib.dump(scaler, 'models/scaler.pkl')
joblib.dump(feature_cols, 'models/feature_cols.pkl')

print(f"Total model-ready features: {len(feature_cols)}")
print("Encoding and scaling completed successfully!")
````

````python

X_train = train_fe[feature_cols]; X_val = val_fe[feature_cols]; X_test = test_fe[feature_cols]

# Classification target
clf_encoder = LabelEncoder()
y_train_clf = clf_encoder.fit_transform(train_fe['emi_eligibility'])
y_val_clf = clf_encoder.transform(val_fe['emi_eligibility'])
y_test_clf = clf_encoder.transform(test_fe['emi_eligibility'])
joblib.dump(clf_encoder, 'models/clf_target_encoder.pkl')
print("Classification classes:", list(clf_encoder.classes_))

# Regression target
y_train_reg = train_fe['max_monthly_emi'].values
y_val_reg = val_fe['max_monthly_emi'].values
y_test_reg = test_fe['max_monthly_emi'].values
print("Regression target range:", y_train_reg.min(), "-", y_train_reg.max())
````

## 🤖 Step 4 — Classification Models: EMI Eligibility
Training 3 required models: **Logistic Regression** (interpretable baseline), **Random Forest** (feature importance), **XGBoost** (high-performance gradient boosting).

````python

mlflow.set_tracking_uri('sqlite:///mlflow.db')
mlflow.set_experiment('EMIPredict_Classification')

clf_models = {
    'Logistic Regression': LogisticRegression(max_iter=500, random_state=RANDOM_STATE, n_jobs=-1),
    'Random Forest': RandomForestClassifier(n_estimators=150, max_depth=14, min_samples_leaf=5,
                                             random_state=RANDOM_STATE, n_jobs=-1),
    'XGBoost': XGBClassifier(n_estimators=200, max_depth=7, learning_rate=0.1, tree_method='hist',
                              objective='multi:softprob', num_class=3, random_state=RANDOM_STATE,
                              eval_metric='mlogloss', n_jobs=-1)
}

clf_results = {}
for name, model in clf_models.items():
    t0 = time.time()
    with mlflow.start_run(run_name=name):
        model.fit(X_train, y_train_clf)
        preds = model.predict(X_val)
        proba = model.predict_proba(X_val)
        acc = accuracy_score(y_val_clf, preds)
        prec = precision_score(y_val_clf, preds, average='weighted')
        rec = recall_score(y_val_clf, preds, average='weighted')
        f1 = f1_score(y_val_clf, preds, average='weighted')
        auc = roc_auc_score(y_val_clf, proba, multi_class='ovr', average='weighted')

        mlflow.log_params(model.get_params())
        mlflow.log_metrics({'accuracy':acc,'precision':prec,'recall':rec,'f1':f1,'roc_auc':auc})
        if name == 'XGBoost':
            mlflow.xgboost.log_model(model, 'model')
        else:
            mlflow.sklearn.log_model(model, 'model')

    clf_results[name] = {'model':model,'accuracy':acc,'precision':prec,'recall':rec,'f1':f1,'roc_auc':auc,
                          'train_time_s': time.time()-t0}
    print(f"{name:22s} | Acc: {acc:.4f} | Prec: {prec:.4f} | Rec: {rec:.4f} | F1: {f1:.4f} | ROC-AUC: {auc:.4f} | {time.time()-t0:.1f}s")
````

## 📈 Step 5 — Regression Models: Maximum Monthly EMI
Training 3 required models: **Linear Regression** (baseline), **Random Forest Regressor**, **XGBoost Regressor**. Target: RMSE < 2000 INR.

````python

mlflow.set_tracking_uri('sqlite:///mlflow.db')
mlflow.set_experiment('EMIPredict_Regression')

reg_models = {
    'Linear Regression': LinearRegression(n_jobs=-1),
    'Random Forest': RandomForestRegressor(n_estimators=150, max_depth=14, min_samples_leaf=5,
                                            random_state=RANDOM_STATE, n_jobs=-1),
    'XGBoost': XGBRegressor(n_estimators=250, max_depth=7, learning_rate=0.08, tree_method='hist',
                             random_state=RANDOM_STATE, n_jobs=-1)
}

reg_results = {}
for name, model in reg_models.items():
    t0 = time.time()
    with mlflow.start_run(run_name=name):
        model.fit(X_train, y_train_reg)
        preds = model.predict(X_val)
        rmse = np.sqrt(mean_squared_error(y_val_reg, preds))
        mae = mean_absolute_error(y_val_reg, preds)
        r2 = r2_score(y_val_reg, preds)
        mape = np.mean(np.abs((y_val_reg - preds) / np.where(y_val_reg==0, 1, y_val_reg))) * 100

        mlflow.log_params(model.get_params())
        mlflow.log_metrics({'rmse':rmse,'mae':mae,'r2':r2,'mape':mape})
        if name == 'XGBoost':
            mlflow.xgboost.log_model(model, 'model')
        else:
            mlflow.sklearn.log_model(model, 'model')

    reg_results[name] = {'model':model,'rmse':rmse,'mae':mae,'r2':r2,'mape':mape,'train_time_s':time.time()-t0}
    print(f"{name:22s} | RMSE: {rmse:.1f} | MAE: {mae:.1f} | R2: {r2:.4f} | MAPE: {mape:.2f}% | {time.time()-t0:.1f}s")
````
